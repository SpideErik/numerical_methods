1. Создайте массив формы (6, 4) из np.arange(24). Выведите .shape, .ndim, .size, .dtype.


```python
import numpy as np

a = np.arange(24).reshape(6, 4)

print('shape: ', a.shape)
print('ndim: ', a.ndim)
print('size: ', a.size)
print('dtype: ', a.dtype)
```

Перестройте его в (4, 6), затем в (2, 3, 4), затем обратно в (24,). Убедитесь через np.array_equal, что значения не изменились.

```python
b = a.copy()
b.reshape(4, 6)
print('reshape(4, 6)\n', b)
b.reshape(2, 3, 4)
print('reshape(2, 3, 4)\n', b)
b.reshape(24)
print('a is b', a is b)
print('np.array_equal', np.array_equal(a, b))

```

Возьмите столбец [:, 2] — это view или копия? Проверьте: измените элемент среза и посмотрите на оригинал.


```python
b = a.copy()
b.reshape(4, 6)
column = b[:,2]
column[0] = 100
print('view Колонка 2:', column)
print('b тоже изменился\n', b)


```

Отфильтруйте элементы > 10 булевой маской. Посчитайте их среднее.


```python
b = a[a>10]
print('a[a>10]', b)

print('среднее', b.mean())

```

Вычтите из каждой строки её среднее (axis=1, keepdims=True) — убедитесь, что результат имеет ту же форму и сумма по строкам ≈ 0.

```python
a.reshape(4, 6)
m = a.mean(axis=1, keepdims=True)
print('Средние значения по строкам\n', m)
a = a - m # a -=m нельзя, потому что среднии имеют тип float
print('a после вычитания среднего по строкам\n', a)
print('shape: ', a.shape)
print('ndim: ', a.ndim)
print('size: ', a.size)
print('dtype: ', a.dtype)

s = a.sum(axis=1)
print('Сумма по строкам\n', s)


```
